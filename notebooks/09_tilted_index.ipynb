**What this notebook does.** Under the factor objective of risk-models-for-index-tracking an optimiser spends a budget of forecast tracking error on exposure to the four targeted factors. The tilted index moves the benchmark in the same direction with no estimate of risk: each industry's weight goes up or down with its score, as far as the mandate's bound on each active weight and the floor at zero allow. This notebook measures what the optimiser's work buys: the tilted index's forecast exposure per basis point of realised tracking error beside each of the seven optimised paths at a budget of 100 basis points a year, month by month from 1979-07 to 2026-08, with a paired block bootstrap and Holm's correction across the seven, and the tilted index's tracking error, turnover, costs, information ratio, realised beta and attribution beside theirs.

**Where the optimised paths come from.** The seven paths, their monthly records and their measures are read from tables of risk-models-for-index-tracking at commit ddbb75c, copied unchanged into `results/202608/source/`, and the code that built them is copied unchanged into `src/bp/source/`, each file with a first line that names its source file and the commit. `tests/test_source_copies.py` checks every copy. Before any figure of the tilted index is computed, section 1 rebuilds one optimised path from the copied code and requires it to reproduce the recorded tracking error within a band of 0.1 basis point a year.

The seven optimised paths, by their labels in the copied tables. Each holds the mandate at a budget of 100 basis points a year, and is labelled "| 100" in the tables below.

| Label | What the optimiser used |
|---|---|
| sample, daily 3y | the sample covariance of the last three years of daily returns, scaled to a month |
| ledoit_wolf, daily 3y | the same, shrunk towards a matrix with the same variance for every industry and no correlation, by the method of Ledoit and Wolf (2004b) |
| factor, daily 3y | the covariance implied by a regression of each industry on the six factors over the same three years |
| pca, daily 3y | the covariance implied by the first five principal components of the same three years |
| sample, monthly 120 | the sample covariance of the last 120 monthly returns |
| robust, all five | no single estimator: the weights whose largest forecast tracking error across the five estimates above is smallest |
| RiskMetrics, daily 5y | J.P. Morgan's 1996 RiskMetrics covariance on five years of daily returns, each day's weight halving every year of trading days |

**How the tilted index is built.** Each month:

1. Start from the month's benchmark weights. Hold Coal, Oil and Util at half their benchmark weight, set Smoke and Guns to zero, and add the weight this removes to the other 44 industries in proportion to their benchmark weights. Call these weights w0.
2. Compute each of the 44 industries' score from that month's betas and centre it on its mean weighted by w0, so that the centred scores z sum to zero under w0.
3. Set the weight of each of the 44 to w0 times (1 + k z), with one k for all 44: the largest value at which every active weight of the 44 stays within 2 percentage points and no weight falls below zero.

The weights still sum to one, because the centred scores sum to zero under w0. An industry whose score is above the w0-weighted mean gets more weight than w0 gives it, and one below gets less. The tilted index starts as the benchmark in 1979-07, as the optimised paths do, rebalances to these weights every month, and its turnover is measured against its own drifted weights. It is not held to the turnover cap or to a beta of one, and its months above the cap and its realised beta are reported. The construction is `tilted_index` in `src/bp/naive.py`.

```python
%%writefile src/bp/naive.py
"""The two naive portfolios under the index mandate, and the measures that set them beside the optimised paths (notebooks 08 and 09).

A naive portfolio takes its weights from the benchmark weights alone, or from the benchmark weights and one score per
industry, with no covariance estimate:

- naive_tilt: Coal, Oil and Util held at half their benchmark weight, and the weight this removes spread over the
  other 46 industries in proportion to their benchmark weights (notebook 08).
- tilted_index: the same tilt with tobacco and weapons excluded, the removed weight spread over the other 44, and each
  of the 44 then scaled by one plus a common multiple k of its centred factor score, k as large as the bound on each
  active weight and the floor at zero allow (notebook 09).

run_path holds a naive portfolio month by month as evaluation.run_path holds an optimised one: the path starts as the
benchmark, rebalances to the month's weights every month, and measures its one-way turnover against its previous weights
drifted by the previous month's returns. The remaining functions compute the measures the two notebooks report.

The exposure that factor_objective measures is the forecast one: the active weights times each industry's betas from the
three years of daily returns before the month. realised_exposure measures it after the fact, from the portfolio's own
monthly active returns regressed on the six factors' monthly returns.
"""
from __future__ import annotations

import numpy as np
import pandas as pd

from . import constants as C
from .source import attribution as A
from .source import evaluation as E
from .source import optimiser as O


def naive_tilt(b: np.ndarray, tilt: np.ndarray) -> tuple[np.ndarray, dict]:
    """The naive tilt's weights for one month, from the benchmark weights b; tilt marks Coal, Oil and Util."""
    w = b.copy()
    removed = (1.0 - C.TILT_MAX_SHARE_OF_BENCHMARK) * b[tilt].sum()
    w[tilt] = C.TILT_MAX_SHARE_OF_BENCHMARK * b[tilt]
    rest = ~tilt
    w[rest] = b[rest] * (1.0 + removed / b[rest].sum())
    return w, {"removed weight": removed}


def tilted_index(b: np.ndarray, tilt: np.ndarray, excluded: np.ndarray, score: np.ndarray,
                 names: list[str]) -> tuple[np.ndarray, dict]:
    """The tilted index's weights for one month.

    b: the benchmark weights; tilt and excluded: the masks of the tilted and the excluded industries; score: each
    industry's score, the sum of its betas to the four targeted factors; names: the industries' names, in b's order.
    Returns the weights and k, the constraint that set it ("bound" or "floor") and the industry at which it binds.
    """
    rest = ~(tilt | excluded)
    w0 = b.copy()
    w0[tilt] = C.TILT_MAX_SHARE_OF_BENCHMARK * b[tilt]
    w0[excluded] = 0.0
    removed = (1.0 - C.TILT_MAX_SHARE_OF_BENCHMARK) * b[tilt].sum() + b[excluded].sum()
    w0[rest] = b[rest] * (1.0 + removed / b[rest].sum())
    s, w_rest, b_rest = score[rest], w0[rest], b[rest]
    z = s - (w_rest @ s) / w_rest.sum()           # the score centred on its w0-weighted mean over the 44
    a = w_rest - b_rest                            # the active weight of w0, zero or more for each of the 44
    c = w_rest * z                                 # the change in active weight per unit of k
    rest_names = [n for n, keep in zip(names, rest) if keep]
    limits = []
    for i in range(len(z)):
        if c[i] > 0:
            limits.append(((C.ACTIVE_WEIGHT_BOUND - a[i]) / c[i], "bound", rest_names[i]))
        elif c[i] < 0:
            limits.append(((a[i] + C.ACTIVE_WEIGHT_BOUND) / -c[i], "bound", rest_names[i]))
            limits.append((1.0 / -z[i], "floor", rest_names[i]))
    k, source, name = min(limits, key=lambda x: x[0])
    w = w0.copy()
    w[rest] = w_rest * (1.0 + k * z)
    return w, {"removed weight": removed, "k": k, "k set by": source, "k set at": name,
               "largest active weight of w0": float(np.abs(w0 - b)[rest].max())}


def run_path(weights_of, months: pd.PeriodIndex, weights: pd.DataFrame, returns_m: pd.DataFrame, betas: dict,
             factors: list[str], sign: np.ndarray) -> pd.DataFrame:
    """One naive path: weights_of(month, b) gives the month's weights and a dict of descriptions.

    Each month records the active return, the one-way turnover of the path and of the benchmark, the largest active
    weight, the effective number and the industries held, the six active exposures with that month's betas, and the
    objective, the exposures summed with the signs in sign. The weights go into the column "weights".
    """
    recs = []
    w_prev, prev = None, None
    for m in months:
        b = weights.loc[m].to_numpy()
        r = returns_m.loc[m].to_numpy()
        w, info = weights_of(m, b)
        if w_prev is None:
            w0, b_drift = b.copy(), b.copy()
        else:
            w0 = O.drifted_weights(w_prev, returns_m.loc[prev].to_numpy())
            b_drift = O.drifted_weights(weights.loc[prev].to_numpy(), returns_m.loc[prev].to_numpy())
        x = A.active_exposures(betas[m], w, b)
        recs.append({"month": m, "active_return": float(w @ r - b @ r), "turnover": O.one_way_turnover(w, w0),
                     "benchmark_turnover": O.one_way_turnover(b, b_drift), "largest active weight": float(np.abs(w - b).max()),
                     "effective_number": O.effective_number(w), "held": int((w > 1e-6).sum()), "objective": float(sign @ x),
                     **{f"exposure {f}": float(v) for f, v in zip(factors, x)}, **info, "weights": w})
        w_prev, prev = w, m
    return pd.DataFrame(recs).set_index("month")


def realised_beta(active, bench_ex) -> float:
    """The slope of the portfolio's monthly excess return (the benchmark's plus the active return) on the benchmark's."""
    bench_ex = np.asarray(bench_ex, dtype=float)
    port_ex = bench_ex + np.asarray(active, dtype=float)
    return float(np.cov(port_ex, bench_ex, ddof=1)[0, 1] / np.var(bench_ex, ddof=1))


def measures(active, turnover, bench_ex, objective=None, exposures=None) -> dict:
    """Realised tracking error, gross active return, turnover, cost and net active return at each cost level, the
    information ratios, the realised beta, and, given the monthly objective and the monthly exposures (a table with a
    column "exposure F" for each factor F), the exposure per basis point and the mean exposure to each factor."""
    active = np.asarray(active, dtype=float)
    te = E.realised_tracking_error(active)
    gross = 12.0 * float(active.mean())
    out = {"realised TE": te, "active return (gross)": gross, "turnover (one-way)": float(np.mean(turnover)),
           "information ratio (gross)": gross / te, "realised beta": realised_beta(active, bench_ex)}
    for c in C.EXT_COST_LEVELS:
        bp = int(round(c * 1e4))
        cost = 12.0 * float(np.mean(turnover)) * c
        out[f"cost at {bp} bp"] = cost
        out[f"net active return at {bp} bp"] = gross - cost
        out[f"information ratio at {bp} bp"] = (gross - cost) / te
    if objective is not None:
        out["mean objective"] = float(np.mean(objective))
        out["exposure per bp of realised TE"] = float(np.mean(objective)) / (1e4 * te)
    if exposures is not None:
        for f in C.ATTRIBUTION_FACTORS:
            out[f"mean exposure {f}"] = float(np.mean(exposures[f"exposure {f}"]))
    return out


def resample_indices(n: int) -> np.ndarray:
    """The circular block resamples of evaluation.block_bootstrap_band: ceil(n / block) block starts per resample, each block
    running past the last month into the first, the generator seeded with EVAL_BOOTSTRAP_SEED."""
    block, draws = C.EVAL_BOOTSTRAP_BLOCK_MONTHS, C.EVAL_BOOTSTRAP_DRAWS
    rng = np.random.default_rng(C.EVAL_BOOTSTRAP_SEED)
    k = int(np.ceil(n / block))
    starts = rng.integers(0, n, size=(draws, k))
    return ((starts[:, :, None] + np.arange(block)[None, None, :]) % n).reshape(draws, k * block)[:, :n]


def exposure_per_bp(objective: np.ndarray, active: np.ndarray) -> np.ndarray:
    """The mean objective over 10,000 times the realised tracking error, along the last axis."""
    return objective.mean(axis=-1) / (1e4 * active.std(axis=-1, ddof=1) * np.sqrt(12.0))


def bootstrap_difference(d_star: np.ndarray) -> tuple[float, float, float]:
    """The 95% percentile interval of the resampled differences and the two-sided p-value: twice the smaller of the shares
    at or below zero and at or above zero, each counted as (count + 1) / (draws + 1), at most 1."""
    tail = 100.0 * (1.0 - C.EVAL_BOOTSTRAP_LEVEL) / 2.0
    lo, hi = np.percentile(d_star, [tail, 100.0 - tail])
    draws = len(d_star)
    p = min(1.0, 2.0 * min((int((d_star <= 0).sum()) + 1) / (draws + 1), (int((d_star >= 0).sum()) + 1) / (draws + 1)))
    return float(lo), float(hi), p


def information_ratios(active, costs: dict, idx: np.ndarray) -> dict:
    """The information ratio gross and net of each cost, each with the standard deviation of the ratio over the resamples
    idx; the cost is held at the full-sample figure, 12 times the mean one-way turnover times the cost level."""
    a = np.asarray(active, dtype=float)
    a_star = a[idx]
    te_star = a_star.std(axis=1, ddof=1) * np.sqrt(12.0)
    g_star = 12.0 * a_star.mean(axis=1)
    te = E.realised_tracking_error(a)
    row = {"information ratio (gross)": 12.0 * a.mean() / te, "standard error (gross)": float((g_star / te_star).std(ddof=1))}
    for c, cost in costs.items():
        row[f"information ratio at {c} bp"] = (12.0 * a.mean() - cost) / te
        row[f"standard error at {c} bp"] = float(((g_star - cost) / te_star).std(ddof=1))
    return row


def information_ratio_star(active, cost: float, idx: np.ndarray) -> np.ndarray:
    """The information ratio on each resample idx, net of the cost per year held at its full-sample figure: the resampled
    ratios whose standard deviation information_ratios reports."""
    a_star = np.asarray(active, dtype=float)[idx]
    return (12.0 * a_star.mean(axis=1) - cost) / (a_star.std(axis=1, ddof=1) * np.sqrt(12.0))


def factor_design(factors) -> np.ndarray:
    """The design of the regression of a monthly active return on the factors: a column of ones, then each factor's
    monthly return in the order given."""
    f = np.asarray(factors, dtype=float)
    return np.column_stack([np.ones(len(f)), f])


def realised_exposure(active, design: np.ndarray, targeted) -> float:
    """The realised exposure to the targeted factors: the sum of their slopes in the least-squares regression of the
    monthly active return on the design. targeted holds the design's column numbers of the targeted factors."""
    b = np.linalg.lstsq(design, np.asarray(active, dtype=float), rcond=None)[0]
    return float(b[list(targeted)].sum())


def realised_exposure_per_bp(active, design: np.ndarray, targeted, idx: np.ndarray | None = None):
    """The realised exposure over 10,000 times the realised tracking error, on the months as they are or, given idx, on each
    resample of them, with the regression run again on the resampled months."""
    a = np.asarray(active, dtype=float)
    if idx is None:
        return realised_exposure(a, design, targeted) / (1e4 * a.std(ddof=1) * np.sqrt(12.0))
    out = np.empty(len(idx))
    for d, j in enumerate(idx):
        out[d] = realised_exposure(a[j], design[j], targeted) / (1e4 * a[j].std(ddof=1) * np.sqrt(12.0))
    return out
```

```python
fo_eval, fo_eval_name = source_table("factor_objective_evaluation")
fo_eval = fo_eval.set_index(fo_eval.columns[0])
fo_month, _ = source_table("factor_objective_monthly")
fo_month["month"] = pd.PeriodIndex(fo_month["month"], freq="M")
MAIN = int(round(1e4 * C.FO_BUDGET_MAIN))
label = f"{C.NAIVE_REPRODUCED} | {MAIN}"
MANDATE = C.CONSTRAINT_SETS[C.FO_CONSTRAINT_SET]
t0 = time.time()
rebuilt = E.run_factor_path(months, Sigma, betas, weights, returns_m, MANDATE, S, C.FO_BUDGET_MAIN, tilt_mask, mandate_mask)
te_rebuilt = E.realised_tracking_error(rebuilt["active_return"])
te_recorded = float(fo_eval.loc[label, "realised TE"])
recorded = fo_month[fo_month["path"] == label].set_index("month").loc[months]
reproduction = pd.DataFrame([{
    "path": label, "committed table": fo_eval_name, "rebuilt realised TE (bp)": 1e4 * te_rebuilt,
    "committed realised TE (bp)": 1e4 * te_recorded, "difference (bp)": 1e4 * (te_rebuilt - te_recorded),
    "largest monthly active return difference": float(np.abs(rebuilt["active_return"].to_numpy() - recorded["active_return"].to_numpy()).max()),
    "largest monthly objective difference": float(np.abs(rebuilt["objective"].to_numpy() - recorded["objective"].to_numpy()).max()),
    "band (bp)": 1e4 * C.ATTRIBUTION_REPRODUCTION_TOL}])
reproduction["passes"] = reproduction["difference (bp)"].abs() <= reproduction["band (bp)"]
reproduction.to_csv(out("tilted_index_reproduction"), index=False, float_format="%.8f")
print(f"rebuilt in {time.time() - t0:.0f}s")
print(reproduction.T.to_string())
assert reproduction["passes"].all(), "the rebuilt path does not reproduce the recorded tracking error, so no figure of the tilted index is computed"
```

## 2. The tilted index, month by month

The cell runs the tilted index over the 566 months and checks each month's weights: they sum to one, none is negative, Coal, Oil and Util are at half their benchmark weight, Smoke and Guns are at zero, and each of the other 44 industries is within 2 percentage points of its benchmark weight. It also reports k: its range over the months, and which limit set it, the bound on an active weight or the floor at zero, at which industry.

```python
score = {m: betas[m] @ S for m in months}
index_path = N.run_path(lambda m, b: N.tilted_index(b, tilt_mask, mandate_mask, score[m], industries),
                        months, weights, returns_m, betas, FACTORS, S)
W = np.vstack(index_path["weights"].to_numpy())
B = weights.loc[months].to_numpy()
free = ~(tilt_mask | mandate_mask)
assert np.abs(W.sum(axis=1) - 1.0).max() < 1e-12 and (W >= -1e-15).all()
assert np.abs(W[:, tilt_mask] - C.TILT_MAX_SHARE_OF_BENCHMARK * B[:, tilt_mask]).max() < 1e-15 and (W[:, mandate_mask] == 0).all()
assert np.abs(W - B)[:, free].max() <= C.ACTIVE_WEIGHT_BOUND + 1e-12
index_path.drop(columns="weights").to_csv(out("tilted_index_monthly"), float_format="%.10f")
pd.DataFrame(W, index=index_path.index, columns=industries).to_csv(out("tilted_index_weights"), float_format="%.10f")
k = index_path["k"]
k_summary = pd.DataFrame([{"k smallest": float(k.min()), "month of the smallest": str(k.idxmin()), "k largest": float(k.max()),
                           "month of the largest": str(k.idxmax()), "k median": float(k.median()),
                           "months set by the floor": int((index_path["k set by"] == "floor").sum()),
                           "months set by the bound": int((index_path["k set by"] == "bound").sum()),
                           "largest active weight of w0": float(index_path["largest active weight of w0"].max())}])
k_summary.to_csv(out("tilted_index_k"), index=False, float_format="%.8f")
k_set_at = index_path["k set at"].value_counts().rename("months")
k_set_at.to_csv(out("tilted_index_k_set_at"))
floor_months = index_path.index[index_path["k set by"] == "floor"]
print(k_summary.T.to_string())
print(k_set_at.to_string())
print("months set by the floor:", ", ".join(sorted(set(index_path.loc[floor_months, "k set at"]))), f"{floor_months.min()} to {floor_months.max()}")
row = N.measures(index_path["active_return"], index_path["turnover"], bench_ex, index_path["objective"], index_path)
row.update({"months above the turnover cap": int((index_path["turnover"] > C.TURNOVER_CAP_MONTHLY_ONE_WAY).sum()),
            "first month turnover": float(index_path["turnover"].iloc[0]), "largest monthly turnover": float(index_path["turnover"].max()),
            "month of the largest turnover": str(index_path["turnover"].idxmax()), "mean benchmark turnover": float(index_path["benchmark_turnover"].mean()),
            "largest active weight over the months": float(index_path["largest active weight"].max()),
            "mean effective number": float(index_path["effective_number"].mean()), "mean industries held": float(index_path["held"].mean())})
measures = pd.DataFrame([row], index=pd.Index(["tilted index"], name="portfolio"))
measures.to_csv(out("tilted_index_measures"), float_format="%.10f")
print(measures.T.to_string())
```

## 3. Forecast exposure per basis point: the tilted index against each optimised path

For each path the cell draws the same 2,000 block bootstrap resamples of the months for the tilted index and the path together, computes the difference in forecast exposure per basis point on each, and reports the 95 per cent percentile interval and a two-sided p-value: twice the smaller of the shares of resampled differences at or below zero and at or above zero, each counted as (count + 1) / (2,000 + 1). Holm's correction is applied across the seven.

```python
o2, a2 = index_path["objective"].to_numpy(), index_path["active_return"].to_numpy()
index_epb = float(N.exposure_per_bp(o2, a2))
index_star = N.exposure_per_bp(o2[IDX], a2[IDX])
rows = []
for v in C.NAIVE_PATHS:
    rec = fo_month[fo_month["path"] == f"{v} | {MAIN}"].set_index("month").loc[months]
    op, ap = rec["objective"].to_numpy(dtype=float), rec["active_return"].to_numpy(dtype=float)
    path_epb = float(N.exposure_per_bp(op, ap))
    lo, hi, p = N.bootstrap_difference(index_star - N.exposure_per_bp(op[IDX], ap[IDX]))
    rows.append({"path": f"{v} | {MAIN}", "tilted index": index_epb, "path value": path_epb,
                 "path, committed table": float(fo_eval.loc[f"{v} | {MAIN}", "exposure per bp of realised TE"]),
                 "difference": index_epb - path_epb, "lower": lo, "upper": hi, "p-value": p})
bootstrap = pd.DataFrame(rows)
bootstrap["Holm p-value"] = PT.holm(bootstrap["p-value"].to_numpy())
bootstrap.to_csv(out("tilted_index_bootstrap"), index=False, float_format="%.10f")
print(bootstrap.to_string(index=False, float_format=lambda x: f"{x:.6f}"))
print(f"statistically significant at the 5 per cent level after Holm's correction: {int((bootstrap['Holm p-value'] < C.NAIVE_SIGNIFICANCE).sum())} of {len(bootstrap)}")
```

## 7. Two checks written after the results were known

Both checks were written on 28 November 2026, once the figures of sections 3 to 6 were known.

The first tests each difference between the tilted index's information ratio and a path's, before costs and at each cost level. The two ratios are measured on the same months, so the difference is computed on each of the 2,000 resamples of section 5, with the cost held at its full-sample figure. Its paired standard error is the standard deviation of the resampled differences, and its interval and p-value follow the rule of section 3, with Holm's correction across the seven paths at each cost level.

The second measures the factor exposure each portfolio realised. The factor objective of sections 3 and 4 is the forecast exposure: each month, the active weights times each industry's betas from the three years of daily returns before the month. The realised exposure is measured after the fact: each portfolio's 566 monthly active returns are regressed, with an intercept, on the six factors' monthly returns, and the slopes on the four targeted factors are added up. Divided by the realised tracking error in basis points, it gives the realised exposure per basis point. Each path's difference from the tilted index is tested on the same 2,000 resamples, with the regression run again on each, by the rule of section 3 and Holm's correction across the seven.

```python
# Added after the run, 28 November 2026: the paired test of the information ratios.
index_costs = cost_of(measures, "tilted index")
rows = []
for label, col, bp in (("gross", "information ratio (gross)", None), ("at 50 bp", "information ratio at 50 bp", 50),
                       ("at 100 bp", "information ratio at 100 bp", 100)):
    index_ir_star = N.information_ratio_star(a2, 0.0 if bp is None else index_costs[bp], IDX)
    block = []
    for v in C.NAIVE_PATHS:
        path = f"{v} | {MAIN}"
        rec = fo_month[fo_month["path"] == path].set_index("month").loc[months]
        d_star = index_ir_star - N.information_ratio_star(rec["active_return"], 0.0 if bp is None else cost_of(fo_eval, path)[bp], IDX)
        lo, hi, p = N.bootstrap_difference(d_star)
        d = information_ratios.loc["tilted index", col] - information_ratios.loc[path, col]
        block.append({"naive portfolio": "tilted index", "path": path, "ratio": label, "naive": information_ratios.loc["tilted index", col],
                      "path value": information_ratios.loc[path, col], "difference": d, "paired standard error": float(d_star.std(ddof=1)),
                      "lower": lo, "upper": hi, "p-value": p})
    block = pd.DataFrame(block)
    block["Holm p-value"] = PT.holm(block["p-value"].to_numpy())
    rows.append(block)
ir_differences = pd.concat(rows, ignore_index=True)
ir_differences.to_csv(out("tilted_index_information_ratio_differences"), index=False, float_format="%.8f")
print(ir_differences.to_string(index=False, float_format=lambda x: f"{x:.4f}"))
print(ir_differences.assign(alone=ir_differences["p-value"] < C.NAIVE_SIGNIFICANCE, holm=ir_differences["Holm p-value"] < C.NAIVE_SIGNIFICANCE)
      .groupby("ratio", sort=False)[["alone", "holm"]].sum().to_string())

# Added after the run, 28 November 2026: the realised exposure to the targeted factors.
design = N.factor_design(P.factors_m.loc[months, FACTORS])
targeted_cols = [1 + FACTORS.index(f) for f in TARGETED]
index_rx = N.realised_exposure(a2, design, targeted_cols)
index_rx_epb = N.realised_exposure_per_bp(a2, design, targeted_cols)
index_rx_star = N.realised_exposure_per_bp(a2, design, targeted_cols, IDX)
rows = [{"portfolio": "tilted index", "forecast exposure": float(o2.mean()), "realised exposure": index_rx,
         "realised TE (bp)": 1e4 * E.realised_tracking_error(a2), "forecast exposure per bp": index_epb, "realised exposure per bp": index_rx_epb,
         "realised over forecast": index_rx / float(o2.mean())}]
for v in C.NAIVE_PATHS:
    path = f"{v} | {MAIN}"
    rec = fo_month[fo_month["path"] == path].set_index("month").loc[months]
    op, ap = rec["objective"].to_numpy(dtype=float), rec["active_return"].to_numpy(dtype=float)
    rx = N.realised_exposure(ap, design, targeted_cols)
    rx_epb = N.realised_exposure_per_bp(ap, design, targeted_cols)
    lo, hi, p = N.bootstrap_difference(index_rx_star - N.realised_exposure_per_bp(ap, design, targeted_cols, IDX))
    rows.append({"portfolio": path, "forecast exposure": float(op.mean()), "realised exposure": rx, "realised TE (bp)": 1e4 * E.realised_tracking_error(ap),
                 "forecast exposure per bp": float(N.exposure_per_bp(op, ap)), "realised exposure per bp": rx_epb, "realised over forecast": rx / float(op.mean()),
                 "difference": index_rx_epb - rx_epb, "lower": lo, "upper": hi, "p-value": p})
realised = pd.DataFrame(rows).set_index("portfolio")
realised["forecast multiple of the tilted index"] = realised["forecast exposure per bp"] / index_epb
realised["realised multiple of the tilted index"] = realised["realised exposure per bp"] / index_rx_epb
realised.loc[realised.index[1:], "Holm p-value"] = PT.holm(realised["p-value"].iloc[1:].to_numpy())
realised.to_csv(out("tilted_index_realised_exposure"), float_format="%.10f")
print(realised.T.to_string(float_format=lambda x: f"{x:.6f}"))
print(f"statistically significant at the 5 per cent level after Holm's correction: {int((realised['Holm p-value'] < C.NAIVE_SIGNIFICANCE).sum())} of {len(C.NAIVE_PATHS)}")
```

## What this notebook establishes, and what it does not settle

The tilted index has a mean forecast exposure of 0.022 at a realised tracking error of 119.3 basis points a year, a forecast exposure per basis point of 0.000185. The seven optimised paths at the budget of 100 basis points have 0.000824 to 0.000979 per basis point, 4.5 to 5.3 times as much, and the tilted index's exposure per basis point is lower than each path's, each difference statistically significant at the 5 per cent level on the paired block bootstrap after Holm's correction across the seven, with adjusted p-values of 0.007. The tilted index's realised tracking error lies inside the paths' range, 115.6 to 150.1 basis points a year, so the comparison is made at a tracking error of the same size. The paths of the five estimators other than RiskMetrics have 0.000478 to 0.000633 per basis point at the budget of 75 basis points and 0.001015 to 0.001073 at 150, both above the tilted index.

The forecast exposure is the quantity the optimiser makes as large as it can, computed with the same estimated betas. Measured on the returns in section 7, a check added after the results were known, the paths realise 0.40 to 0.56 of their forecast exposure and the tilted index 1.9 times its own. On realised exposure per basis point the paths obtain 1.06 to 1.48 times the tilted index's 0.000351, and no difference is statistically significant at the 5 per cent level after Holm's correction, with adjusted p-values of 1.000. The 95 per cent intervals of the differences are too wide to rule out a realised multiple of about 2.2 to 2.7. The paths' residual of -28.4 to -37.4 basis points a year in the attribution, against the tilted index's -1.4, is consistent with a forecast exposure above the realised one in months when the targeted factors earned a premium.

The tilted index's exposure is small partly because of its form. One k scales all 44 industries, and the change in an industry's active weight is its weight times k times its centred score, so the largest industries reach the bound of 2 percentage points first and set k for all the others: k is set by the bound in 549 of the 566 months, most often at Banks (184 months) and Softw, computer software (138), and by the floor at zero in the other 17, all at Gold, from 1981-04 to 1984-01. Its median over the months is 0.206. This is a limit of the construction, and no other construction of a tilted index is run here.

The tilted index trades less than the paths: 0.77% of the portfolio a month against 1.78% to 2.00%, and more than the cap of 2% a month in 38 of the 566 months, 14% of it in its first month. Its realised beta is 1.004, against 0.999 to 1.012 for the paths.

Over the sample the tilted index earned an active return of +16.4 basis points a year before costs, against +2.3 to +15.5 for the paths, and +7.2 after a cost of 100 basis points per unit of turnover, against -20.0 to -5.9. No verdict was set on returns. On the paired test of section 7, added after the results were known, no difference in information ratio is statistically significant at the 5 per cent level at any cost: before costs the differences are 0.02 to 0.12 with paired standard errors of 0.12 to 0.15. These figures therefore do not show that the tilted index earns more than an optimised path, or that it earns less. After costs the gaps widen because the tilted index trades less than every path, a property of the two methods.

**What this notebook does not settle.**

- The tilted index is one way to move the benchmark towards the targeted factors with no estimate of risk. A construction that scaled each industry's active weight by its score, or that capped k industry by industry, would have a different exposure, and none is run here.
- The figures are those of one sample and one vintage, 202608. A later vintage reproduces the comparisons, and not every digit, because French revises history.
- The optimised paths are read from risk-models-for-index-tracking at commit ddbb75c and one of them is rebuilt here. The others are taken as recorded, and their checks are in that repository.
